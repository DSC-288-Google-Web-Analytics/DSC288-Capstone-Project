```python
import pandas as pd
import numpy as np
import json
import matplotlib.pyplot as plt
```

```python
def feature_value_count(dataset, column):
    return dataset[column].value_counts()

def features_remove(dataset, columns=['sessionId', 
                                      'socialEngagementType', 
                                      'visitId', 
                                      'customDimensions',
                                      'hits'
                                     ]):
    return dataset.drop(columns=columns, errors='ignore')

def features_type_correct(x):
    try:
        return json.loads(x) if isinstance(x, str) else {}
    except:
        return {}

def split_json_features(df):
    df = features_remove(df)
    
    devices = df['device'].apply(features_type_correct).apply(pd.Series)
    device = devices[['isMobile', 'deviceCategory']]

    geos = df['geoNetwork'].apply(features_type_correct).apply(pd.Series)
    geo = geos[["subContinent", "country"]]

    totals = df['totals'].apply(features_type_correct).apply(pd.Series)
    totals = totals.rename(columns={'hits': 'hit'})
    total = total = totals[[
        "visits",
        "hit",
        "pageviews",
        "transactions",
        "transactionRevenue"
    ]].apply(pd.to_numeric, errors='coerce') 

    traffics = df['trafficSource'].apply(features_type_correct).apply(pd.Series)
    traffic = traffics[['source', 'medium']]
    
    df_new = pd.concat(
        [df.drop(columns=['device', 'geoNetwork', 'totals', 'trafficSource']),
         device, geo, total, traffic],
        axis=1
    )
    
    return df_new
```

```python
def is_revenue(dataset):
    df = dataset.copy()
    df['transactionRevenue'] = pd.to_numeric(df['transactionRevenue'], errors='coerce').fillna(0)
    df['transactions'] = pd.to_numeric(df['transactions'], errors='coerce').fillna(0)
    df['has_revenue'] = (df['transactionRevenue'] > 0).astype(int)

    return df
```

```python
revenue_dataset = is_revenue(processing_data)
revenue_dataset.head()
```

```python
print(revenue_dataset['has_revenue'].mean())
```

After processing the nested JSON fields, we discovered that `transactionRevenue` contained a large number of missing values ​​(NaN). This is not data loss, but rather indicates that no transaction was generated during that access. Therefore, we filled it with 0 and constructed a binary label `has_revenue` for the classification task.

```python
def type_change(df):
    df['visitNumber'] = pd.to_numeric(df['visitNumber'], errors='coerce')
    df['visitStartTime'] = pd.to_datetime(df['visitStartTime'], unit='s')

    df['date'] = pd.to_datetime(df['date'], format='%Y%m%d')
    df['year'] = df['date'].dt.year
    df['month'] = df['date'].dt.month
    df['day'] = df['date'].dt.day

    return df
```

```python
revenue_dataset = type_change(revenue_dataset)
revenue_dataset.head()
```

```python
import matplotlib.pyplot as plt

def draw_features_hist(df, column, top_n=10):
    value_counts = df[column].value_counts()
    if len(value_counts) > top_n:
        top_counts = value_counts.head(top_n)
    else:
        top_counts = value_counts
    total = value_counts.sum()
    proportions = (top_counts / total * 100)
    print(f"\nTop {len(top_counts)} categories in {column}:")
    print(proportions.round(2).astype(str) + "%")
    top_counts.plot(kind='bar')

    plt.title(f'Top {len(top_counts)} {column} Distribution')
    plt.xlabel(column)
    plt.ylabel('Count')
    plt.xticks(rotation=45)
    plt.show()
```

1. Key Findings: Highly Concentrated "Lightweight" Access
- The data exhibits a clear long-tail distribution (Power Law Distribution).
- Low-frequency access dominates: Users with only one click (1.0 hit) account for a staggering 49.43%, nearly half of the total.
- Rapid churn: Users with 3 or fewer clicks account for over 72.5% cumulatively. This indicates that the vast majority of users leave after only a very limited interaction upon entering the website.

2. Conversion Funnel and User Engagement Analysis  
- The funnel decays extremely rapidly: From 1 click to 2 clicks, the number of users experiences a "cliff-like" drop (from 49.43% to 15.38%). This means that the attractiveness of the website's homepage or landing page is crucial for user retention.
- High-engagement users are scarce: Users with 10 clicks account for only 1.12%. Typically, in e-commerce or content-based data, this small minority of high-click users contributes the majority of conversion revenue or activity.

```python
visit_conversion = df.groupby('visitNumber')['has_revenue'].mean()
visit_conversion = visit_conversion[visit_conversion > 0]
print(visit_conversion)
visit_conversion.plot()

plt.title('Conversion Rate by Visit Number')
plt.xlabel('Visit Number')
plt.ylabel('Conversion Rate')
plt.show()
```

Data analysis reveals that the website's users exhibit a clear "deliberate" consumption pattern. The conversion rate for initial visits (visitNumber=1) is extremely low, at only 0.4%, indicating that users rarely make impulsive purchases. However, as the number of visits increases, the conversion rate shows a significant step-like increase: the conversion rate for users with 2-10 repeat visits rises to 2%-4%, and when the number of visits exceeds 30, the conversion rate for some high-frequency visitors even exceeds 10%-20%. This trend suggests that user engagement is the core driver of conversion, with early multiple visits primarily serving as a comparison and evaluation phase. Therefore, marketing strategies should shift from simply "acquiring new users" to "remarketing" based on repeat visit frequency, shortening the decision-making path for high-frequency visitors through targeted incentives.

## Comprehensive Summary of EDA on the Google Analytics Dataset
1. Data Overview and Quality Assessment

- Raw Data Structure: The dataset contains highly nested JSON fields (device, geoNetwork, totals, trafficSource), which have been successfully flattened into tabular features using custom functions.

- Handling of Missing Values:

- - `transactionRevenue` exhibits an extremely high missing rate (>99%); it was confirmed that `NaN` signifies the absence of a transaction. These values ​​have been imputed with 0, and a binary label `has_revenue` has been constructed.


- Removal of Constant Features: Fields such as `socialEngagementType` consist entirely of a single unique value; as they contribute no information to the model, they were removed during the preprocessing phase.

2. Key Business Insights

- Traffic Distribution (Channel Distribution):

- - Organic Search (approx. 42%) serves as the primary traffic source, followed by Social and Direct channels.

- - The distribution of traffic sources is highly consistent between the training and test sets, providing a solid foundation for model generalization.

- User Interaction Behavior (Hits Analysis):

- - Extreme Concentration: Approximately 49.4% of visits generate only a single "hit."

- - High Drop-off Rate: Over 72% of users drop off within three clicks. This suggests that the retention capability of the Landing Page constitutes a critical bottleneck in efforts to boost conversion rates.

- Conversion Performance:

- - Extreme Class Imbalance: The site-wide conversion rate stands at a mere 0.936%.

- - Subsequent modeling efforts must prioritize strategies such as undersampling, oversampling, or adjusting loss function weights to address this imbalance.

3. Recommendations for Subsequent Modeling

- Feature Selection: Prioritize features such as `hits`, `pageviews`, and `visitNumber`, as they directly reflect user intent.


- Evaluation Metrics: Due to the class imbalance, Accuracy should not be used as an evaluation metric; instead, RMSE (for predicting Log Revenue) or AUC-ROC / F1-score (for predicting conversion probability) are recommended.
